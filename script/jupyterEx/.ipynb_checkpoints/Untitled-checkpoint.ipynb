```python
b_scan_plot = []
for i in range(130):
    aProfile = np.load("/home/hui/gr-sfcwRadar/gui/scan_"+str(i)+".npy")
    b_scan_plot.append(np.abs(aProfile))
plt.imshow(np.array(b_scan_plot).T,aspect = 'auto', origin="lower")
```

```python
from sklearn import decomposition
M = np.array(b_scan_plot)
u, s, v = decomposition.randomized_svd(np.abs(M), 1)
low_rank = u @ np.diag(s) @ v
plt.figure()
plt.imshow(low_rank, cmap='jet')
plt.figure()
plt.imshow(M-low_rank, cmap='jet')
```
