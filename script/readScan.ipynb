```python
import numpy as np
import matplotlib.pyplot as plt
import json
```

```python
b_scan = []
dir = "/mnt/others/rangeProfile/"
for i in range(900):
    fileName = dir+"scan_" + str(i) + ".npy"
    #print(fileName)
    with open(fileName,"r") as file:
        jsonData = json.load(file)
        imag = np.asarray(jsonData['rangeProfile_imag'])
        real = np.asarray(jsonData['rangeProfile_real'])
        result = 1j*imag
        result += real
        rangeProfile = 10*np.log10(np.abs(result)**2)
        #rangeProfile = np.abs(result)
        b_scan.append(rangeProfile)

b_scan_np = np.array(b_scan)
#plt.imshow(b_scan_np,cmap="hot",vmin=60,vmax=340,aspect="auto")
cal = np.mean(b_scan_np[0:20,:],axis=0)
b_scan_cal = b_scan_np-cal
plt.imshow(b_scan_cal,cmap="hot",aspect="auto")
plt.figure()
plt.imshow(b_scan_np,cmap="hot",aspect="auto")
```
